```python
def U_htoff(qc: Circuit, index=0, cx = False, beam = False):
    # qc.reset((1,2))
    if cx and beam:
        with open("htoffcxbeam64_15_1e-2.txt", "r") as f:
            htoff_decomp = [
                [tuple(item.split("_")[0:1] + [int(x) for x in item.split("_")[1:]])
                for item in line.strip().split(",")]
                for line in f
            ]
    elif cx:
        with open("htoffcx15_1e-2.txt", "r") as f:
            htoff_decomp = [
                [tuple(item.split("_")[0:1] + [int(x) for x in item.split("_")[1:]])
                for item in line.strip().split(",")]
                for line in f
            ]
    else:
        with open("htoff15_1e-2.txt", "r") as f:
                htoff_decomp = [
                    [tuple(item.split("_")[0:1] + [int(x) for x in item.split("_")[1:]])
                    for item in line.strip().split(",")]
                    for line in f
                ]
    # htoff_decomp.reverse()
    for i in htoff_decomp[index]:
        if i[0] == "X":
            qc.X(i[1])
        elif i[0] == "Z":
            qc.Z(i[1])
        elif i[0] == "H":
            qc.H(i[1])
        elif i[0] == "CCX":
            qc.CCX(i[1],i[2],i[3])
        elif i[0] == "CX":
            qc.CX(i[1],i[2])
        else:
            print("MISSING GATE")
```

```python
for i in range(7):
    circ = Circuit(3)

    U_htoff(circ, index=2*i+1, cx=True, beam=True)

    print("before:", circ.n_gates)

    GATES = {OpType.H, OpType.X, OpType.Z, OpType.CX, OpType.CCX}

    RepeatPass(SequencePass([CommuteThroughMultis(), RemoveRedundancies()])).apply(circ)

    print("after:", circ.n_gates, circ.get_commands())
```
